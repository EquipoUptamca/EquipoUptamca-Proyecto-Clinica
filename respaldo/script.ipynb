# [dbo].[Chatbot_Reports]

```sql
/****** Object:  View [dbo].[Chatbot_Reports]    Script Date: 20/6/2026 3:03:38 p. m. ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE VIEW [dbo].[Chatbot_Reports] AS
SELECT 
    ci.user_type,
    ci.response_type,
    COUNT(*) as interaction_count,
    AVG(CASE WHEN ci.success = 1 THEN 1.0 ELSE 0.0 END) * 100 as success_rate,
    MAX(ci.timestamp) as last_interaction,
    u.nombre_completo as user_name
FROM Chatbot_Interactions ci
LEFT JOIN Usuarios u ON ci.user_id = u.id_usuario
GROUP BY ci.user_type, ci.response_type, u.nombre_completo
GO
```

# [dbo].[CleanupChatbotData]

```sql
/****** Object:  StoredProcedure [dbo].[CleanupChatbotData]    Script Date: 20/6/2026 3:03:38 p. m. ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[CleanupChatbotData]
    @retention_days INT = 90
AS
BEGIN
    SET NOCOUNT ON;
    
    DELETE FROM Chatbot_Interactions 
    WHERE timestamp < DATEADD(day, -@retention_days, GETDATE())
    
    SELECT @@ROWCOUNT as 'deleted_rows'
END
GO
```

# [dbo].[GetChatbotStats]

```sql
/****** Object:  StoredProcedure [dbo].[GetChatbotStats]    Script Date: 20/6/2026 3:03:38 p. m. ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[GetChatbotStats]
    @days INT = 7,
    @user_type NVARCHAR(20) = NULL
AS
BEGIN
    SET NOCOUNT ON;
    
    SELECT 
        COUNT(*) as total_interactions,
        SUM(CASE WHEN success = 1 THEN 1 ELSE 0 END) as successful_interactions,
        AVG(CASE WHEN success = 1 THEN 1.0 ELSE 0.0 END) * 100 as success_rate,
        COUNT(DISTINCT user_id) as unique_users,
        user_type
    FROM Chatbot_Interactions 
    WHERE timestamp >= DATEADD(day, -@days, GETDATE())
    AND (@user_type IS NULL OR user_type = @user_type)
    GROUP BY user_type
    ORDER BY total_interactions DESC
END
GO
```
